```python
import pandas as pd
from IPython.display import display

def show_first_10_rows(data):
    """Convert tabular data to a DataFrame and display its first 10 rows."""
    if isinstance(data, pd.Series):
        data = data.to_frame()
    elif not isinstance(data, pd.DataFrame):
        data = pd.DataFrame(data)

    with pd.option_context(
        "display.max_columns", None,
        "display.max_rows", 10,
        "display.width", None,
        "display.max_colwidth", None
    ):
        display(data.head(10))

    return data.head(10)

# Replace `data` with your DataFrame, list, dictionary, or other tabular data.
data = pd.read_csv('tickets_human_handled_updated.csv')
first_10_rows = show_first_10_rows(data)
```
